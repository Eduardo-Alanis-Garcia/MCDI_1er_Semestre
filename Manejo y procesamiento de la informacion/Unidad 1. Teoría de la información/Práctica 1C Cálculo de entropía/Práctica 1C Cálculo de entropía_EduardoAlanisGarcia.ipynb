### Función leer archivos

```python
def leer_archivo(nombre_archivo, direccion=direccion):
  ruta = os.path.join(direccion, nombre_archivo)

  with open(ruta, "r", encoding="utf-8") as f:
      texto = f.read()

  return texto
```

### Función de quitar acentos

```python
import unicodedata

def remove_accents(input_str):
  nfkd_form = unicodedata.normalize('NFKD', input_str)
  return u"".join([c for c in nfkd_form if not unicodedata.combining(c)])
```

### Caracteres a eliminar

```python
caracteres_eliminar = ';:,.\\-"\'/()[]¿?¡!{}~<>|«»-—’'
```

### Función preprocesar texto

```python
def preprocesar_texto(texto):
  texto = texto.lower()  # Minusculas
  texto = remove_accents(texto)  # Acentos
  texto = texto.replace('\t', ' ').replace('\n', ' ')  # Tabulaciones y saltos de linea

  for c in caracteres_eliminar:
    texto = texto.replace(c, '')  # Eliminar caracteres mencionados

  return texto
```

### Funciones de cálculo de frecuencias

```python
from collections import Counter

def calcular_frecuencias_caracteres(texto):
  frecuencias = Counter(texto)
  ordenado = frecuencias.most_common()

  caracteres = []
  valores = []

  for c, f in ordenado:
    if c == ' ':
      caracteres.append('espacio')
    else:
      caracteres.append(c)

    valores.append(f)

  guardar = {"caracteres": caracteres, "valores": valores}

  return guardar
```

```python
def calcular_frecuencias_palabras(texto):
  palabras = texto.split()
  frecuencias = Counter(palabras)
  ordenado = frecuencias.most_common()

  palabras = []
  valores = []

  for palabra, frecuencia in ordenado:
      palabras.append(palabra)
      valores.append(frecuencia)

  guardar = {"palabras": palabras, "valores": valores}

  return guardar
```

## Entropía
Es una medida de la incertidumbre de una variable aleatoria X. Mide el grado de sorpresa al ocurrir un evento. Mientras más raro el evento ocurrido, mayor será la sorpresa. La entropía de una variable aleatoria discreta se define como

$$H(X) = -\sum_{x\in X} p(x)\cdot\log_2 p(x)$$

donde $p(x)$  es la función de probabilidad; en computación se usa $log$ base $2$  y así, la entropía dice el menor número de bits que se necesitan para mandar un mensaje. En la definición anterior usamos la convención $0\cdot\log(0)=0$.

```python
import math

def calcular_entropia(valores):
  total = sum(valores)
  entropia = 0

  for f in valores:
    p = f / total
    if p > 0:
      entropia += p * math.log2(p)

  return -entropia
```

## Ejercicio 1: Calcula la entropía global a nivel de carácter de los documentos text_1 al text_5 de manera independiente. No tomar en cuenta el espacio para el cálculo de la entropía.

```python
resultados_caracteres = {}

for archivo in archivos_texto:
  texto_original = leer_archivo(nombre_archivo=archivo)
  texto_procesado = preprocesar_texto(texto=texto_original)
  frecuencias = calcular_frecuencias_caracteres(texto=texto_procesado)

  # Quitar el espacio antes de calcular la entropia
  caracteres = frecuencias["caracteres"]
  valores = frecuencias["valores"]


  valores_sin_espacio = []

  for i in range(len(caracteres)):
    if caracteres[i] != "espacio":
      valores_sin_espacio.append(valores[i])

  entropia = calcular_entropia(valores_sin_espacio)

  resultados_caracteres[archivo] = {
      "texto_procesado": texto_procesado,
      "frecuencias": frecuencias,
      "valores_sin_espacio": valores_sin_espacio,
      "entropia": entropia
  }

  print(f"{archivo} tiene entropía = {entropia:} bits "
        f"donde los símbolos distintos sin espacio son {len(valores_sin_espacio)} "
        f"y el total de caracteres son {sum(valores_sin_espacio)}")
```

## Ejercicio 2: Calcula la entropía global a nivel de palabra de los documentos libro_1 y libro_2 de manera independiente, primero sin quitar stopwords y luego quitándolas.

### Funcion para quitar stopwords

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
_STOPWORDS = stopwords.words("spanish")

# quitar stopwords
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in _STOPWORDS:
      continue
    else:
      new_text.append(word)
  return " ".join(new_text)
```

```python
resultados_palabras = {}

for archivo in archivos_libros:
  texto_original = leer_archivo(nombre_archivo=archivo)
  texto_procesado = preprocesar_texto(texto=texto_original)

  # Sin quitar stopwords
  frecuencias_con_stopwords = calcular_frecuencias_palabras(texto_procesado)
  entropia_con_stopwords = calcular_entropia(frecuencias_con_stopwords["valores"])

  # Quitando stopwords
  texto_sin_stopwords = remove_stopwords(texto_procesado)
  frecuencias_sin_stopwords = calcular_frecuencias_palabras(texto_sin_stopwords)
  entropia_sin_stopwords = calcular_entropia(frecuencias_sin_stopwords["valores"])

  resultados_palabras[archivo] = {
      "texto_procesado": texto_procesado,
      "frecuencias_con_stopwords": frecuencias_con_stopwords,
      "entropia_con_stopwords": entropia_con_stopwords,
      "texto_sin_stopwords": texto_sin_stopwords,
      "frecuencias_sin_stopwords": frecuencias_sin_stopwords,
      "entropia_sin_stopwords": entropia_sin_stopwords
  }

  print(f"{archivo}")
  print(f"Palabras totales (con stopwords): {len(texto_procesado.split())}, "
        f"palabras únicas: {len(frecuencias_con_stopwords['valores'])},",
        f"entropía con stopwords: {entropia_con_stopwords:} bits")

  print(f"Palabras totales (sin stopwords): {len(texto_sin_stopwords.split())}, "
        f"palabras únicas: {len(frecuencias_sin_stopwords['valores'])}, "
        f"entropía sin stopwords: {entropia_sin_stopwords:} bits")
  print()
```
